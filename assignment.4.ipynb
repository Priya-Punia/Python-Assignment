1. Explain Pandas and its role in Data Analysis and Machine Learning

Pandas is an open-source Python library used for data manipulation and analysis. It provides powerful data structures such as Series (1-dimensional) and DataFrame (2-dimensional) that make it easy to store, organize, and process data.

Role of Pandas in Data Analysis
Loads data from files such as CSV, Excel, and SQL databases.
Cleans data by handling missing values and duplicates.
Filters, sorts, and groups data efficiently.
Performs statistical analysis and generates summaries.
Helps in data visualization by working with libraries like Matplotlib and Seaborn.
Role of Pandas in Machine Learning
Prepares and preprocesses data before training models.
Handles missing values and categorical variables.
Selects relevant features and transforms data into suitable formats.
Splits and organizes datasets for training and testing.
Works seamlessly with machine learning libraries such as Scikit-learn.
Common Functions in Pandas
pd.read_csv() – Load data from a CSV file.
head() – Display the first few rows.
info() – Show dataset information.
describe() – Generate summary statistics.
isnull() – Detect missing values.
fillna() – Fill missing values.

1. Difference between Series and DataFrame
Series is a one-dimensional data structure that stores a single column of data.
DataFrame is a two-dimensional data structure that stores data in rows and columns.
A Series can contain only one type of information, whereas a DataFrame can contain multiple columns with different types of data.
Series is created using pd.Series(), while DataFrame is created using pd.DataFrame().
Example: A list of student marks is a Series, while a table containing Name, Age, and Marks is a DataFrame.
2. Difference between loc[] and iloc[]
loc[] is used for label-based indexing, whereas iloc[] is used for integer position-based indexing.
In loc[], rows and columns are selected using their names or labels.
In iloc[], rows and columns are selected using index numbers.
Syntax of loc[] is df.loc[row_label, column_label].
Syntax of iloc[] is df.iloc[row_index, column_index]

```python
#3. Create a Pandas DataFrame using the dataset 
import pandas as pd

# Create DataFrame
data = {
    "Name": ["Amit", "Neha", "Rahul", "Sneha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

# Display complete DataFrame
print("Original DataFrame:")
print(df)

# Display first 2 rows
print("\nFirst 2 Rows:")
print(df.head(2))

# Display only Name and Marks columns
print("\nName and Marks Columns:")
print(df[["Name", "Marks"]])

# Add a new column named 'Result'
df["Result"] = ["Pass", "Pass", "Pass", "Pass"]

print("\nDataFrame after adding Result column:")
print(df)

# Display students with marks greater than 80
print("\nStudents with Marks greater than 80:")
print(df[df["Marks"] > 80])
```

b) Expected Concepts Used
pd.DataFrame() → To create a DataFrame.
head() → To display the first few rows.
Column Selection → df[["Name", "Marks"]].
Conditional Filtering → df[df["Marks"] > 80].

```python
#4. Create a DataFrame and Handle Missing Values
import pandas as pd
import numpy as np

# Creating the DataFrame
data = {
    "Name": ["A", "B", "C", "D"],
    "Age": [25, np.nan, 27, 24],
    "Salary": [30000, 40000, np.nan, 35000]
}

df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame:")
print(df)
```

```python
#a) Tasks
# 1. Identify missing values
print("\nMissing Values:")
print(df.isnull())
#2. Fill missing Age with average age
avg_age = df["Age"].mean()
df["Age"] = df["Age"].fillna(avg_age)
#3. Fill missing Salary with median salary
median_salary = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(median_salary)
#4. Display cleaned dataset
print("\nCleaned DataFrame:")
print(df)
```
